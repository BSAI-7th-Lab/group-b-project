## Preliminary smoke test (2 epochs)

```python
from ultralytics import YOLO
import torch

print("CUDA Available:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0))

# Load pretrained YOLOv8 nano model
model = YOLO("yolov8n.pt")

# Short training test
results = model.train(
    data=str(prepared_yaml.resolve()),
    epochs=2,
    imgsz=640,
    batch=16,
    device=0,
    workers=0,
    seed=42,
    project=str((output_root.parent / "YOLO_Experiments").resolve()),
    name="baseline_smoke_test",
    exist_ok=False,
    plots=True
)
```

# Baseline YOLOv8 training

```python
from ultralytics import YOLO

baseline_model = YOLO("yolov8n.pt")

baseline_results = baseline_model.train(
    data=str(prepared_yaml.resolve()),
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    workers=0,
    seed=42,
    deterministic=True,

    optimizer="AdamW",
    lr0=0.001,
    momentum=0.9,
    weight_decay=0.0005,

    patience=50,
    close_mosaic=10,
    amp=True,

    project=str((output_root.parent / "YOLO_Experiments").resolve()),
    name="YOLOv8n_Baseline_50_Epochs",
    exist_ok=False,
    plots=True
)
```

```python
from pathlib import Path
import json

baseline_dir = (
    output_root.parent
    / "YOLO_Experiments"
    / "YOLOv8n_Baseline_50_Epochs"
)

baseline_best = baseline_dir / "weights" / "best.pt"

assert baseline_best.exists(), "Baseline best.pt not found!"

baseline_metrics = {
    "model": "YOLOv8n",
    "experiment": "Baseline",
    "epochs": 50,
    "image_size": 640,
    "batch_size": 16,
    "optimizer": "AdamW",
    "learning_rate": 0.001,
    "momentum": 0.9,
    "weight_decay": 0.0005,
    "seed": 42,
    "evaluation_split": "validation",
    "precision": 0.877,
    "recall": 0.800,
    "mAP50": 0.897,
    "mAP50_95": 0.451,
    "weights_path": str(baseline_best.resolve())
}

metrics_path = baseline_dir / "baseline_metrics.json"

with open(metrics_path, "w", encoding="utf-8") as f:
    json.dump(baseline_metrics, f, indent=4)

print("Baseline metrics saved:", metrics_path)
print("Best model exists:", baseline_best.exists())
print(json.dumps(baseline_metrics, indent=4))
```

## Baseline-only test evaluation
The following code is adapted from original cell index 33 to exclude PSO evaluation. It has not been executed in this extracted notebook.

```python
experiments_root = output_root.parent / "YOLO_Experiments"
baseline_path = experiments_root / "YOLOv8n_Baseline_50_Epochs" / "weights" / "best.pt"
baseline_test_model = YOLO(str(baseline_path))
baseline_test = baseline_test_model.val(data=str(prepared_yaml.resolve()), split="test", imgsz=640, batch=16, device=0, workers=0, plots=True, project=str(experiments_root), name="Baseline_Test_Evaluation", exist_ok=False)
for label, value in [("Precision", baseline_test.box.mp), ("Recall", baseline_test.box.mr), ("mAP50", baseline_test.box.map50), ("mAP50-95", baseline_test.box.map)]:
    print(f"{label}: {value:.4f}")
```
